```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import folium
from haversine import haversine
```

```python
# Create a map centered around New York City
map_nyc = folium.Map(location=[40.7128, -74.0060], zoom_start=10)

# Add pickup locations to the map
for _, row in data.iterrows():
    folium.CircleMarker(
        location=[row['pickup_latitude'], row['pickup_longitude']],
        radius=5,
        color='blue',
        fill=True,
        fill_color='blue'
    ).add_to(map_nyc)

# Save the map to an HTML file
map_nyc.save('pickup_locations.html')
```

```python
# Define the coordinates of major airports
jfk_coords = (40.6413, -73.7781)
ewr_coords = (40.6925, -74.1687)
lga_coords = (40.7770, -73.8726)

# Calculate distances to each airport
data['distance_to_jfk'] = data.apply(lambda row: haversine((row['pickup_latitude'], row['pickup_longitude']), jfk_coords), axis=1)
data['distance_to_ewr'] = data.apply(lambda row: haversine((row['pickup_latitude'], row['pickup_longitude']), ewr_coords), axis=1)
data['distance_to_lga'] = data.apply(lambda row: haversine((row['pickup_latitude'], row['pickup_longitude']), lga_coords), axis=1)

# Create a scatter plot for pickup locations with distances to airports
plt.figure(figsize=(12, 8))
plt.scatter(data['pickup_longitude'], data['pickup_latitude'], c=data['distance_to_jfk'], cmap='viridis', alpha=0.6)
plt.colorbar(label='Distance to JFK (miles)')
plt.title('Pickup Locations with Distance to JFK')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()

plt.figure(figsize=(12, 8))
plt.scatter(data['pickup_longitude'], data['pickup_latitude'], c=data['distance_to_ewr'], cmap='viridis', alpha=0.6)
plt.colorbar(label='Distance to EWR (miles)')
plt.title('Pickup Locations with Distance to EWR')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()

plt.figure(figsize=(12, 8))
plt.scatter(data['pickup_longitude'], data['pickup_latitude'], c=data['distance_to_lga'], cmap='viridis', alpha=0.6)
plt.colorbar(label='Distance to LGA (miles)')
plt.title('Pickup Locations with Distance to LGA')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
